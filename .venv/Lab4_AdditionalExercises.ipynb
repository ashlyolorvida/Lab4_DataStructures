```python
import random
from collections import defaultdict

def generate_and_organize_data(last_name, seed_num, favorite_artist):
    print("=" * 60)
    print(f" ENGINEERING TEST DATA ORGANIZER")
    print("=" * 60)
    print(f"Inputs Profile:")
    print(f"  Last Name: {last_name}")
    print(f"  Seed Number: {seed_num}")
    print(f"  Favorite Artist: {favorite_artist}\n")

    # -------------------------------------------------------------------------
    # RULE 1: Generate Test Sources & Categories
    # -------------------------------------------------------------------------
    sources = [f"{last_name}_T{i}" for i in range(1, 6)]
    categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

    # Seed the random number generator using the seed_num and character positions/values
    # to ensure consistency while incorporating the character data.
    artist_char_sum = sum(ord(char) * (idx + 1) for idx, char in enumerate(favorite_artist))
    combined_seed = seed_num + artist_char_sum
    random.seed(combined_seed)

    # -------------------------------------------------------------------------
    # RULE 2 & 3: Generate Base Measurement Values & Records
    # -------------------------------------------------------------------------
    original_records = []
    
    # Generate standard distinct records
    for src in sources:
        for cat in categories:
            # Generate a realistic-looking engineering measurement value
            val = round(random.uniform(10.0, 500.0) + (combined_seed % 10), 2)
            original_records.append((src, cat, val))

    # RULE 4: Include one repeated test-source/category record
    repeated_src = sources[0]
    repeated_cat = categories[0]
    repeated_val = round(original_records[0][2] + random.uniform(1.0, 5.0), 2)
    original_records.append((repeated_src, repeated_cat, repeated_val))

    # RULE 5: Include multiple measurements under at least two categories
    # Adding extra measurements for sources to simulate multiple test runs
    extra_runs = [
        (sources[1], categories[1], round(random.uniform(50.0, 150.0), 2)),
        (sources[2], categories[2], round(random.uniform(10.0, 50.0), 2))
    ]
    original_records.extend(extra_runs)

    # -------------------------------------------------------------------------
    # ASSESSMENT 1: Generated Test Data (Original)
    # -------------------------------------------------------------------------
    print("--- 1. Generated Test Data (Original Dataset) ---")
    print(f"{'Source':<15} | {'Category':<15} | {'Recorded Value':<10}")
    print("-" * 48)
    for src, cat, val in original_records:
        print(f"{src:<15} | {cat:<15} | {val:<10}")
    print(f"Total records generated: {len(original_records)}\n")

    # -------------------------------------------------------------------------
    # ASSESSMENT 2 & 3: Organized Data & Distinct/Repeated Analysis
    # -------------------------------------------------------------------------
    # Grouping by Source and Category
    organized_data = defaultdict(list)
    key_counts = defaultdict(int)

    for src, cat, val in original_records:
        organized_data[(src, cat)].append(val)
        key_counts[(src, cat)] += 1

    print("--- 2 & 3. Organized, Distinct, and Repeated Data Analysis ---")
    print(f"{'Source & Category pair':<30} | {'Status':<10} | {'Recorded Values'}")
    print("-" * 65)
    
    for (src, cat), values in organized_data.items():
        status = "Repeated" if key_counts[(src, cat)] > 1 else "Distinct"
        values_str = ", ".join(map(str, values))
        print(f"{f'{src} - {cat}':<30} | {status:<10} | [{values_str}]")
    print("\n")

    # -------------------------------------------------------------------------
    # ASSESSMENT 4 & 5: Numerical/Categorical Summaries & Ordered Results
    # -------------------------------------------------------------------------
    print("--- 4 & 5. Summarized and Ordered Results (by Category) ---")
    print(f"{'Category':<15} | {'Count':<6} | {'Min':<8} | {'Max':<8} | {'Average':<8}")
    print("-" * 55)
    
    # Sort categories alphabetically for a meaningful ordered layout
    for cat in sorted(categories):
        cat_values = [val for _, c, val in original_records if c == cat]
        if cat_values:
            count = len(cat_values)
            minimum = min(cat_values)
            maximum = max(cat_values)
            average = round(sum(cat_values) / count, 2)
            print(f"{cat:<15} | {count:<6} | {minimum:<8} | {maximum:<8} | {average:<8}")
    print("\n")

    # -------------------------------------------------------------------------
    # ASSESSMENT 6 & 7 / FINAL OUTPUT: Consolidated Summary Dataset
    # -------------------------------------------------------------------------
    print("=" * 60)
    print(" FINAL OUTPUT: CONSOLIDATED SUMMARY TEST DATASET")
    print("=" * 60)
    print(f"{'Source':<15} | {'Category':<15} | {'Runs':<6} | {'Final Processed Value (Avg)':<10}")
    print("-" * 65)
    
    # Sort alphabetically by source then category
    for (src, cat), values in sorted(organized_data.items()):
        avg_val = round(sum(values) / len(values), 2)
        print(f"{src:<15} | {cat:<15} | {len(values):<6} | {avg_val:<10}")
    print("=" * 60)

# -------------------------------------------------------------------------
# Execution / Student Input Prompt
# -------------------------------------------------------------------------
if __name__ == "__main__":
    STUDENT_LAST_NAME = "OLORVIDA"
    STUDENT_SEED_NUM = 261850
    STUDENT_FAVORITE_ARTIST = "NIKI"

    generate_and_organize_data(STUDENT_LAST_NAME, STUDENT_SEED_NUM, STUDENT_FAVORITE_ARTIST)
```

```python
# Student Performance Data Analyzer
# Tailored for 1st-Year Computer Science / Engineering Students

def calculate_score(seed, char):
    """Generates a pseudo-random score between 40 and 100 based on a seed number and character."""
    # A simple formula using ASCII value of the character and the seed number
    score = (seed + ord(char)) % 61 + 40  
    return score

def get_performance_category(score):
    """Determines the performance category based on the score."""
    if score is None or score == "N/A":
        return "Incomplete"
    elif score >= 90:
        return "Excellent"
    elif score >= 75:
        return "Good"
    elif score >= 50:
        return "Pass"
    else:
        return "Fail"

def main():
    print("=== Student Performance Data Analyzer Inputs ===")
    last_name = input("Enter your Last Name: ").strip().upper()
    seed_num = int(input("Enter a Seed Number (integer): "))
    favorite_artist = input("Enter your Favorite Artist: ").strip().upper()
    
    # 5 Common Subjects as required
    subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]
    
    # 1. Generate Individualized Student Records
    raw_records = []
    artist_len = len(favorite_artist)
    
    # Generate 5 student names using LAST_NAME + label
    student_names = [f"{last_name}_{i+1}" for i in range(5)]
    
    # Create baseline records mapping each student to each subject
    char_index = 0
    for student in student_names:
        for subject in subjects:
            # Pick a character from favorite_artist for score generation
            char = favorite_artist[char_index % artist_len] if artist_len > 0 else 'A'
            score = calculate_score(seed_num, char)
            
            raw_records.append({
                "student": student,
                "subject": subject,
                "score": score
            })
            char_index += 1

    # Rule Exception: Include exactly ONE repeated student-subject record
    if len(raw_records) > 0:
        repeated_record = raw_records[0].copy()
        # Alter the score slightly to simulate a re-assessment or duplicate entry
        if repeated_record["score"] is not None:
            repeated_record["score"] = min(100, repeated_record["score"] + 5)
        raw_records.append(repeated_record)

    # Rule Exception: Include exactly ONE student record with a missing score (Incomplete)
    if len(raw_records) > 1:
        raw_records[1]["score"] = "N/A"

    print("\n--- [1] Generated Raw Records ---")
    for r in raw_records:
        print(f"Student: {r['student']} | Subject: {r['subject']} | Score: {r['score']}")

    # 2 & 3. Organize, Clean, and Categorize Data
    cleaned_records = []
    repeated_records_log = []
    incomplete_records_log = []
    
    # Tracking duplicates using a primitive unique tracker key "student_subject"
    seen_student_subjects = []

    for r in raw_records:
        student = r["student"]
        subject = r["subject"]
        score = r["score"]
        
        # Check for Incomplete
        if score == "N/A":
            incomplete_records_log.append(r)
            category = get_performance_category(score)
            cleaned_records.append({**r, "category": category})
            continue
            
        # Check for Repeated/Duplicate entry
        unique_key = f"{student}_{subject}"
        if unique_key in seen_student_subjects:
            repeated_records_log.append(r)
            # Choose to keep the highest or latest score (here we update the existing entry)
            for clean_r in cleaned_records:
                if clean_r["student"] == student and clean_r["subject"] == subject:
                    clean_r["score"] = max(clean_r["score"], score)
                    clean_r["category"] = get_performance_category(clean_r["score"])
        else:
            seen_student_subjects.append(unique_key)
            category = get_performance_category(score)
            cleaned_records.append({**r, "category": category})

    # 4. Present Flagged Anomaly Summaries
    print("\n--- [2] Identified Anomalies ---")
    print(f"Repeated Records Found: {len(repeated_records_log)}")
    for r in repeated_records_log:
        print(f"  -> Duplicate entry found for {r['student']} in {r['subject']}")
        
    print(f"Incomplete Records Found: {len(incomplete_records_log)}")
    for r in incomplete_records_log:
        print(f"  -> Missing grade for {r['student']} in {r['subject']}")

    # 5. Performance Summaries (Calculate Averages per Subject)
    print("\n--- [3] Performance Summaries By Subject ---")
    subject_totals = {sub: 0 for sub in subjects}
    subject_counts = {sub: 0 for sub in subjects}
    
    for r in cleaned_records:
        if isinstance(r["score"], (int, float)):
            subject_totals[r["subject"]] += r["score"]
            subject_counts[r["subject"]] += 1
            
    for sub in subjects:
        count = subject_counts[sub]
        avg = subject_totals[sub] / count if count > 0 else 0
        print(f"Subject: {sub:12} | Valid Entries: {count} | Class Average: {avg:.2f}")

    # 6 & 7. Arrange and Present Consolidated Report (Sorted alphabetically by student name)
    # Simple Bubble Sort to arrange results cleanly without complex libraries
    n = len(cleaned_records)
    for i in range(n):
        for j in range(0, n-i-1):
            if cleaned_records[j]["student"] > cleaned_records[j+1]["student"]:
                cleaned_records[j], cleaned_records[j+1] = cleaned_records[j+1], cleaned_records[j]

    print("\n" + "="*65)
    print(" FINAL CONSOLIDATED STUDENT PERFORMANCE REPORT")
    print("="*65)
    print(f"{'STUDENT NAME':<15} | {'SUBJECT':<12} | {'SCORE':<6} | {'STATUS / CATEGORY'}")
    print("-"*65)
    for r in cleaned_records:
        score_str = str(r['score'])
        print(f"{r['student']:<15} | {r['subject']:<12} | {score_str:<6} | {r['category']}")
    print("="*65)

if __name__ == "__main__":
    main()
```

```python
# Student-Specific Inputs
LAST_NAME = "Olorvida"
SEED_NUM = 26-1850
FAVORITE_ARTIST = "Niki"  

# ----------------------------------------------------------------------
# 1. GENERATE TRANSACTION DATA (Data Generation Rules)
# ----------------------------------------------------------------------

# Rule 1: Generate 5 customer names using LAST_NAME and labels C1 to C5
customers = [f"{LAST_NAME}_{label}" for label in ["C1", "C2", "C3", "C4", "C5"]]

# Rule 2 & 4: Products and fixed prices
product_prices = {
    "NOTEBOOK": 5.00,
    "PEN": 1.50,
    "USB": 15.00,
    "HEADPHONES": 50.00,
    "CALCULATOR": 25.00
}
products_list = list(product_prices.keys())

# Rule 3, 5 & 6: Generate transaction records with quantities & a repeated purchase
# We use the character positions (indices) of FAVORITE_ARTIST and SEED_NUM to derive quantities
raw_transactions = []
artist_len = len(FAVORITE_ARTIST)

for i in range(artist_len):
    # Map each character index to a customer and a product deterministically
    cust_idx = i % len(customers)
    prod_idx = (i + SEED_NUM) % len(products_list)
    
    customer = customers[cust_idx]
    product = products_list[prod_idx]
    
    # Rule 3: Quantity derived from position and SEED_NUM
    quantity = ((i + SEED_NUM) % 4) + 1  # Results in a quantity between 1 and 4
    
    raw_transactions.append({
        "customer": customer,
        "product": product,
        "quantity": quantity
    })

# Rule 6: Guarantee at least one explicit repeated purchase by adding a duplicate
raw_transactions.append({
    "customer": customers[0],
    "product": products_list[0],
    "quantity": 2
})


# ----------------------------------------------------------------------
# 2. ORGANIZE AND ANALYZE DATA (Requirements 2, 3, 4, 5, 6)
# ----------------------------------------------------------------------

organized_records = []
total_dataset_value = 0.0

# Track distinct counts and repeated purchases
unique_customer_products = set()
repeated_purchases = []
distinct_customers = set()
distinct_products = set()

# Summaries storage
customer_summary = {cust: 0.0 for cust in customers}
product_summary = {prod: {"quantity": 0, "revenue": 0.0} for prod in products_list}

for tx in raw_transactions:
    cust = tx["customer"]
    prod = tx["product"]
    qty = tx["quantity"]
    
    # Requirement 3: Calculate transaction value
    unit_price = product_prices[prod]
    tx_value = qty * unit_price
    total_dataset_value += tx_value
    
    # Track distinct elements
    distinct_customers.add(cust)
    distinct_products.add(prod)
    
    # Requirement 4: Identify repeated purchases
    pair = (cust, prod)
    if pair in unique_customer_products:
        if pair not in repeated_purchases:
            repeated_purchases.append(pair)
    else:
        unique_customer_products.add(pair)
        
    # Requirement 5: Aggregate summaries
    customer_summary[cust] += tx_value
    product_summary[prod]["quantity"] += qty
    product_summary[prod]["revenue"] += tx_value
    
    # Save organized details
    organized_records.append({
        "customer": cust,
        "product": prod,
        "quantity": qty,
        "value": tx_value
    })

# Requirement 6: Arrange the analyzed results (Sort records by Customer name)
organized_records.sort(key=lambda x: x["customer"])


# ----------------------------------------------------------------------
# 3. PRESENT CONSOLIDATED REPORT (Requirement 7 / Assessment Data)
# ----------------------------------------------------------------------

print("=" * 65)
print("             FINAL CONSOLIDATED TRANSACTION REPORT             ")
print("=" * 65)
print(f"Student Inputs: Name Reference = {LAST_NAME} | Seed = {SEED_NUM} | Artist = {FAVORITE_ARTIST}\n")

# Assessment 2: Organized Transaction Data
print("1 & 2. ORGANIZED TRANSACTION RECORDS")
print(f"{'Customer':<15} | {'Product':<12} | {'Quantity':<8} | {'Total Value ($)':<12}")
print("-" * 65)
for rec in organized_records:
    print(f"{rec['customer']:<15} | {rec['product']:<12} | {rec['quantity']:<8} | ${rec['value']:>10.2f}")
print("-" * 65)

# Assessment 3: Repeated/Distinct Transactions
print("\n3. REPEATED & DISTINCT TRANSACTION SUMMARY")
print(f"Total Distinct Customers: {len(distinct_customers)}")
print(f"Total Distinct Products:  {len(distinct_products)}")
print("Repeated Customer-Product Pairs Identified:")
if repeated_purchases:
    for cust, prod in repeated_purchases:
        print(f"  - {cust} purchased {prod} multiple times.")
else:
    print("  - None")

# Assessment 5: Customer Summaries
print("\n4. CUSTOMER REVENUE SUMMARY")
print(f"{'Customer':<15} | {'Total Spent ($)':<15}")
print("-" * 35)
for cust, spent in sorted(customer_summary.items()):
    print(f"{cust:<15} | ${spent:>12.2f}")

# Assessment 6: Product Summaries
print("\n5. PRODUCT PERFORMANCE SUMMARY")
print(f"{'Product':<12} | {'Units Sold':<10} | {'Total Revenue ($)':<15}")
print("-" * 45)
for prod, metrics in sorted(product_summary.items()):
    print(f"{prod:<12} | {metrics['quantity']:<10} | ${metrics['revenue']:>12.2f}")

# Assessment 8: Overall Summary
print("=" * 65)
print(f"TOTAL DATASET TRANSACTION VALUE: ${total_dataset_value:,.2f}")
print("=" * 65)
```
